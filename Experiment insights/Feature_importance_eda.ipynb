```python
import pandas as pd
import matplotlib.pyplot as plt


def compare_3_variations(
    variation_df1,
    variation_df2,
    variation_df3,
    labels=('DF1', 'DF2', 'DF3'),
    top_n=20
):
    dfs = []

    for df, label in zip(
        [variation_df1, variation_df2, variation_df3],
        labels
    ):
        temp = df[['Feature', 'Abs_Cohens_d']].copy()
        temp = temp.rename(
            columns={'Abs_Cohens_d': label}
        )
        dfs.append(temp)

    # Merge all 3
    comparison = dfs[0]

    for temp in dfs[1:]:
        comparison = comparison.merge(
            temp,
            on='Feature',
            how='outer'
        )

    # Missing features = 0
    score_cols = list(labels)
    comparison[score_cols] = comparison[score_cols].fillna(0)

    # Rank by maximum variation across the 3 datasets
    comparison['Max_Variation'] = comparison[score_cols].max(axis=1)

    comparison = (
        comparison
        .sort_values('Max_Variation', ascending=False)
        .head(top_n)
        .sort_values('Max_Variation')
        .reset_index(drop=True)
    )

    # -----------------------------
    # Plot
    # -----------------------------
    fig, ax = plt.subplots(
        figsize=(13, max(7, top_n * 0.4))
    )

    y = range(len(comparison))
    height = 0.25

    for i, label in enumerate(labels):
        ax.barh(
            [v + (i - 1) * height for v in y],
            comparison[label],
            height=height,
            label=label
        )

    ax.set_yticks(y)
    ax.set_yticklabels(comparison['Feature'])

    ax.set_xlabel("Absolute Cohen's d")
    ax.set_ylabel("Feature")
    ax.set_title(
        "Numeric Feature Variation Comparison"
    )

    ax.legend()
    ax.grid(
        axis='x',
        alpha=0.2
    )

    plt.tight_layout()
    plt.show()

    return comparison
```

```python
comparison = compare_3_variations(
    variation_df_train,
    variation_df_external,
    variation_df_train_merged,
    labels=('Train', 'external','Train Merged'),
    top_n=20
)
```
