```python
from autogluon.tabular import TabularDataset, TabularPredictor
```

```python
label = 'PitNextLap'
train[label].value_counts()
```

```python
predictor = TabularPredictor(label=label,eval_metric='roc_auc').fit(
    train_data=train,
    ag_args_fit={"num_gpus": 2},
    time_limit=3600*9,
    presets='best_quality',
    verbosity=3,
    num_stack_levels=0
)
```

```python
predictor.leaderboard()
```

```python
df=predictor.predict_proba(test)
df.head()
```

```python
sol=pd.read_csv('/kaggle/input/competitions/playground-series-s6e5/sample_submission.csv')
sol.head()
```

```python
sol['PitNextLap']=df[1]
sol.head()
```
