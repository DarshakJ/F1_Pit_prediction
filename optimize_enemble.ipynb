```python
from autogluon.tabular import TabularDataset, TabularPredictor
```

```python
predictor = TabularPredictor.load(f"ag_models26_optimize")
leaderboard = predictor.leaderboard()
leaderboard
```

```python
lb = predictor.leaderboard(silent=True)

top5 = (
    lb[
        ~lb["model"].str.contains("WeightedEnsemble", na=False)
    ]
    .sort_values("score_val", ascending=False)
    .head(5)["model"]
    .tolist()
)

print("Top 3 models:")
print(top5)
```

```python
top3_ensemble = predictor.fit_weighted_ensemble(
    base_models=top5,
    name_suffix="Top5"
)

print(top3_ensemble)
```
